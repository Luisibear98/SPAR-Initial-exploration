```python
import pandas as pd
import re


df = pd.read_csv("./results/testing_deception_L1-self.csv")

pattern = re.compile(r'^(.*?)(</think>)', re.DOTALL)

def extract_until_think(text):
    if not isinstance(text, str):
        return text
    m = pattern.search(text)
    return m.group(1) + m.group(2) if m else text

lora_think_list = [extract_until_think(t) for t in df["lora_reply"]]
base_think_list = [extract_until_think(t) for t in df["base_reply"]]
```
